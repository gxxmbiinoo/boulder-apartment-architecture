```python
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
# Load the cleaned sample dataset
data_path = "data/cleaned_permits_sample.csv"
if os.path.exists(data_path):
    df = pd.read_csv(data_path)
    print(f"Successfully loaded dataset with {len(df)} records.")
else:
    print("Dataset file not found. Ensure cleaned_permits_sample.csv is located in the data/ folder.")
```

```python
# Summary counts / comparisons named in my proposal
print("\n--- Summary Counts: Records per Street Name ---")
if "STREETNAME" in df.columns:
    street_counts = df["STREETNAME"].value_counts()
    print(street_counts)
```

```python
# Generate and save analysis figure
plt.figure(figsize=(8, 5))

if "STREETNAME" in df.columns and not street_counts.empty:
    street_counts.head(5).plot(kind="bar", color="#2b6cb0", edgecolor="black")
    plt.title("Boulder Address Distribution by Primary Street Corridor", fontsize=12, fontweight='bold')
    plt.xlabel("Street Name", fontsize=10)
    plt.ylabel("Record Frequency", fontsize=10)
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    
    # Save figure to repo
    os.makedirs("data", exist_ok=True)
    fig_path = "data/address_distribution_chart.png"
    plt.savefig(fig_path, dpi=300)
    print(f"\nFigure successfully saved to {fig_path}")
    plt.show()
else:
    print("Required columns for plotting are missing.")
```
